### Writing a file

writing ("w") a file erases the old contents of the file, and replaces it with new contents, if we want to append to old content then use "a"

```python
with open("demo.txt", "w") as f:
    f.write("Assalamualikum\n")
    f.write("Ibrahim\n")
    f.write("Mohammad")
```

```python
with open("demo.txt", "r") as f:
    print(f.read())
```

```python
with open("demo.txt", "r") as f:
    print(f.readline())
```

```python
with open("demo.txt", "r") as f:
    print(f.readlines())
```

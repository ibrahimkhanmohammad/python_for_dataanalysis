### Reading File

```python
f = open("hello.txt", "r")
x = f.read() # x is a str where the txt file is stored in the variable x in str format
print(type(x))
print(x)
f.close()
```

```python
# as x is a str so we can iterate through each ch of that str and solve problems like count numbers, spaces, capital letters etc

f = open("hello.txt", "r")
x = f.read()

for ch in x:
    print(ch, end="")
    
f.close()
```

```python
f = open("hello.txt", "r")
print(f.read(5)) # it prints the first 5 charecters of txt file 
print(f.read(5)) # then next it prints next 5 charecters of txt file
f.close()
```

```python
# now if we read the entire file and then try to read the 5 charecters of that same file then;

f = open("hello.txt", "r")
print(f.read())
print(f.read(5))
print(f.read(15))
f.close()
```

readline method returns each line eachtime

```python
f = open("hello.txt", "r")
print(f.readline())
print(f.readline())
f.close()
```

readlines method returns every line in a list format

```python
f = open("hello.txt", "r")
print(f.readlines())
f.close()
```

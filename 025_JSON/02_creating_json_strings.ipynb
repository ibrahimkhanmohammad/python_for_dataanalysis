### json.dumps() — Python object to JSON string

```python
import json

student = { "name": "Luke",
            "age": 26,
            "marks": [78, 86, 91] }

text = json.dumps(student, indent=2)

print(text)
print(text[2:6])
print(type(text))
```

What's happening?



Python uses True.



JSON uses true.



The result of json.dumps() is a string.

### json.loads() — JSON string to Python object

```python
import json

text = '{ "name": "Mark", "age": 25, "marks": [85, 95, 90] }'

data = json.loads(text)

print(data)
print(data["marks"])
print(type(data))
```

What's happening?


text is a string containing JSON.



json.loads(text) converts it into a Python dictionary.



You can now access its values using dictionary keys.

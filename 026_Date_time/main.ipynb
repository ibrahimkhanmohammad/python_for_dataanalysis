### Date & Time

```python
# Print today's date in DD-MM-YYYY format

import datetime

today = datetime.date.today()
print(today.strftime('%d-%m-%Y'))
```

```python
# Create a date for 15 August 2027 and print the weekday name (for example, Monday or Tuesday)

import datetime

day = datetime.date(2027, 8, 15)
print(day.strftime('%A'))
```

```python
# Convert the string '21-10-2026' into a datetime object and print it

date = '21-10-2026'

import datetime

now = datetime.datetime.strptime(date, '%d-%m-%Y')
print(now)
```

```python
# Calculate the date 30 days from today

import datetime

today = datetime.date.today()
future_date = today + datetime.timedelta(days = 30)
print(today)
print(future_date)
```

```python
# Calculate how many days remain between today and 1 January 2027

import datetime

today = datetime.date.today()
date = datetime.date(2027, 1, 1)
difference = date - today

print(f"No.of days remaining are {difference.days}. ")
```

```python
import datetime

now =  datetime.datetime.now()
print(f"Hour: {now.hour}")
print(f"Minute: {now.minute}")
print(f"Second: {now.second}")
```

```python
import datetime

earlier_time = datetime.datetime(2026, 10, 21, 9, 0, 0)
later_time = datetime.datetime(2026, 10, 21, 14, 0, 0)

difference = later_time - earlier_time

difference = difference.total_seconds() / 3600

print(difference)
```
